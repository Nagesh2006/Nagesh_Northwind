```python
employee_order_count = order_finalData.groupby(['EmployeeID', 'LastName', 'FirstName'])['OrderID'].nunique().reset_index()
employee_order_count = employee_order_count.rename(columns={'OrderID': 'Number_of_Orders'})

highest_orders = employee_order_count.sort_values(by='Number_of_Orders', ascending=False).head(1)
lowest_orders = employee_order_count.sort_values(by='Number_of_Orders', ascending=True).head(1)

print("Employee with the highest number of orders:\n", highest_orders)
print("\nEmployee with the lowest number of orders:\n", lowest_orders)
plt.figure(figsize=(10, 6))
sns.barplot(
    x='Number_of_Orders',
    y='FirstName',
    hue='FirstName',
    data=employee_order_count.sort_values(by='Number_of_Orders'),
    palette=sns.color_palette("rocket", n_colors=len(employee_order_count)),
    legend=False,

)
plt.title('Employee-wise Number of Orders')
plt.xlabel('Number of Orders')
plt.ylabel('Employee')
plt.show()
```
